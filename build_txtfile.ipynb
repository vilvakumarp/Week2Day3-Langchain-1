```python
from langchain_huggingface import HuggingFaceEmbeddings
# Our Small knowledge base ( a list of sentences) to test the embeddings
text1 = [
    "Python is a programming language.",
    "Streamlit is used to build web apps.",
    "Dogs are friendly animals.",
    "Cats like to sleep a lot."
]
#Free, local embeddings model for generating embeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
print("Ready")
```

```python
from langchain_community.vectorstores import FAISS
# Create a FAISS vector store from the text1 and embeddings
vector_store = FAISS.from_texts(text1, embeddings)
print("All texts are stored in the FAISS vector store.")
```

```python
query = "Tell me about ai"
results = vector_store.similarity_search(query, k=2)  # Retrieve top 2 similar texts
print("Top matches for:", query)

for r in results:
    print("-", r.page_content)
```

```python
vector_store.save_local("my_store")
print("Vector store saved to 'my_store'.")
#to load it back later
vector_store = FAISS.load_local("my_store", embeddings, allow_dangerous_deserialization=True)
print("Vector store loaded from 'my_store'.")
```

```python
# Retriever
from langchain_community.vectorstores import FAISS
text2 = [
    "Python is a programming language used for AI.",
    "Streamlit is used to build web apps quickly.",
    "Dogs are friendly and loyal animals.",
    "Langchain helps connect AI model with your data."
]
#Store
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
vector_store = FAISS.from_texts(text2, embeddings)
print("vector_store is ready")
#retrieve 2 nearest content from the vector store
retriever = vector_store.as_retriever(search_kwargs={"k": 2})  # Retrieve top 2 similar texts
print("Retriever is ready")

#question to AI
question = "What is used to build web apps?"
results = retriever.invoke(question)
print("Question:", question)
print("\nTop matches:")

for i, result in enumerate(results, start=1):
    print(f"{i}. {result.page_content}")
```
